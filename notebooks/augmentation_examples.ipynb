```python
import sys
import os

# import from parent directory
module_path = os.path.abspath(os.path.join('..'))

if module_path not in sys.path:
    sys.path.append(module_path)

import torchvision.transforms.v2.functional as TF
import dataset as ds
import tools.utils as utils
import matplotlib.pyplot as plt
```

```python
dataset_path = '/home/aislab/Desktop/tomatoes-segmentation/datasets/ood-enhanced-tomato-greenhouse-dataset/'

annotations_path = os.path.join(dataset_path, 'annotations.json')
images_path = os.path.join(dataset_path, 'images')
masks_path = os.path.join(dataset_path, 'masks')

train, valid, test = utils.get_dataset_splits(os.path.abspath('.'), annotations_path, (1, 0, 0))
```

```python
cols, rows = 5, 4
aspect = 360 / 640

state = {'current_epoch': 1, 'total_epochs': 100}
dataset = ds.SimpleDataset(annotations_path, images_path, masks_path, train, 0, 360, 640, state=state, augment=True, curriculum=True)

fig, axes = plt.subplots(rows, cols, figsize=(15, 15 * aspect * rows / cols))

for i, ax in enumerate(axes.flat):
    img, _ = dataset[0]

    ax.imshow(TF.to_pil_image(img))
    ax.axis('off')

plt.subplots_adjust(wspace=0.1, hspace=0.1)
plt.show()
```

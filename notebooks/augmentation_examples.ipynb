```python
import os
import torchvision.transforms.v2.functional as TF
from dataset import SimpleDataset
import tools.utils as utils
```

```python
dataset_path = '/home/giovanni/Desktop/agricultural-robotics/datasets/enhanced-tomato-greenhouse-dataset/'

annotations_path = os.path.join(dataset_path, 'annotations.json')
images_path = os.path.join(dataset_path, 'images')

train, valid, test = utils.get_dataset_splits('splits.json', annotations_path, (1, 0, 0))
```

```python
import matplotlib.pyplot as plt

cols, rows = 5, 4
aspect = 540 / 960

state = {'current_epoch': 1, 'total_epochs': 100}
dataset = SimpleDataset(annotations_path, images_path, train, 0, 960, 540, state=state, augment=True, curriculum=True)

fig, axes = plt.subplots(rows, cols, figsize=(15, 15 * aspect * rows / cols))

for i, ax in enumerate(axes.flat):
    img, _ = dataset[0]

    ax.imshow(TF.to_pil_image(img))
    ax.axis('off')

plt.subplots_adjust(wspace=0.1, hspace=0.1)
plt.show()
```
